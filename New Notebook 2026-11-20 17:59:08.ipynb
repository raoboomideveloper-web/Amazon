```python
data = [
    (1, "John", 25000),
    (2, "David", 35000),
    (3, "Smith", 45000)
]

df = spark.createDataFrame(
    data,
    ["id", "name", "salary"]
)

df.createOrReplaceTempView("employees")

result = spark.sql("""
    SELECT *
    FROM employees
    WHERE salary > 30000
""")

display(result)
```
