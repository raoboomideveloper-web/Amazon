```python
%sql
select * from employee
```

```python
df = spark.table("employee")

df.show(3)
```

```python
from pyspark.sql.functions import col

df.orderBy(col("ID").desc()).show(3)
```

```python
from pyspark.sql.functions import col

df.select("id", "name").show()
```

```python
from pyspark.sql.functions import col

df.filter(col("id") > 2).orderBy(col("id").desc()).show(2)
```

```python
from pyspark.sql.functions import col
df.filter(col("ID")>2).orderBy(col("ID").desc()).show(2)
```

```python
from pyspark.sql.functions import col
df.filter(col("ID")==1).show(2)
```

```python
from pyspark.sql.functions import col
df.dropDuplicates().show()
```

```python
from pyspark.sql.functions import col
print("total records :",df.count())
```

```python
from pyspark.sql.functions import col

df.filter(col("ID").isNotNull()).count()
```

```python
from pyspark.sql.functions import col
df.select(col("ID")).show()
```

```python
from pyspark.sql.functions import col
df.select(col("ID")).distinct().count()
```

```python
from pyspark.sql.functions import avg

df.groupBy("DEPT").agg(avg("AGE")).show()
```

```python
from pyspark.sql.functions import max

df.groupBy("DEPT").agg(max("AGE")).show()
```
